```python
# Getting customer input

# Getting transaction amount for the month
print('Please enter monthlyTransaction: ')
monthlyTransaction = int(input())

# Getting service fee
serviceFee = 0

if monthlyTransaction != 0:
    print('Please enter account opening year: ')
    accountOpeningYear = int(input())    
    if accountOpeningYear <= 2010:
        serviceFee = 0.015 * monthlyTransaction
    else:
        serviceFee = 0.019 * monthlyTransaction   
    if serviceFee >= 1000:
        serviceFee = 1000    
else:
    serviceFee = 5
    
# Define a formatter
formatter = "{n:,.1f}"
# Apply formatter for the print out
print('The service fee is $' + formatter.format(n=serviceFee))
```
