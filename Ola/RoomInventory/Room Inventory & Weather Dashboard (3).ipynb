```python
""
import pandas as pd  # Import the Pandas library for data manipulation

class RoomInventory:
    def __init__(self):
        # Use pandas to read the CSV file and load it into a DataFrame
        self.hotel_df = pd.read_csv("Clark_County_Room_Inventory_Dec_2023.csv")  # Load the data directly from the CSV file

    def display_inventory(self):
        # Display the first few rows of the DataFrame to show room inventory
        print("Room Inventory Data (First 5 rows):")
        print(self.hotel_df.head())

    def search_hotel(self, hotel_name):
        # Search for the hotel by name using a helper function
        result = self.case_insensitive_search(hotel_name)
        if not result.empty:
            print("Hotel Search Results:")
            print(result[['Property Name', 'Address', 'Phone', 'Rooms']])
        else:
            print("No hotel found with that name.")

    def case_insensitive_search(self, hotel_name):
        # Perform a case-insensitive search for the hotel name
        return self.hotel_df[self.hotel_df['Property Name'].str.contains(hotel_name, case=False, na=False)]

    def check_room_availability(self, hotel_name, rooms_needed):
        # Search for hotel
        hotel = self.case_insensitive_search(hotel_name)
        
        if hotel.empty:
            print(f"No hotel found with name: {hotel_name}")
        else:
            # Check room inventory availability, ensuring 'Rooms' is a numerical type
            available_rooms = hotel['Rooms'].values[0]
            if available_rooms >= rooms_needed:
                print(f"Rooms available in {hotel_name}: {available_rooms}")
                print("Reservation can be made.")
            else:
                print(f"Not enough rooms available in {hotel_name}. Needed: {rooms_needed}, Available: {available_rooms}")

    def generate_room_inventory_report(self):
        # Generate a summary report of total rooms available and average rooms per property
        total_rooms = self.hotel_df['Rooms'].sum()
        average_rooms = self.hotel_df['Rooms'].mean()
        print("Room Inventory Summary Report:")
        print(f"Total number of rooms available: {total_rooms}")
        print(f"Average number of rooms per property: {average_rooms:.2f}")

        # Display the top 5 properties with the most rooms
        top_properties = self.hotel_df.nlargest(5, 'Rooms')
        print("Top 5 Properties with Most Rooms:")
        print(top_properties[['Property Name', 'Rooms']])

# Main menu function to display options and link to room inventory functions
def main_menu():
    # Initialize the RoomInventory class
    room_inventory = RoomInventory()

    while True:
        print("Room Inventory Main Menu")
        print("1. Display Room Inventory")
        print("2. Search for a Hotel")
        print("3. Check Room Availability")
        print("4. Generate Room Inventory Report")
        print("5. Exit")

        try:
            menu_input = int(input("Please enter menu selection (1 - 5): "))
        except ValueError:
            print("Invalid Entry. Please enter a number between 1 and 5.")
            continue

        if menu_input == 1:
            room_inventory.display_inventory()
        elif menu_input == 2:
            hotel_name = input("Enter hotel name to search: ").strip()
            room_inventory.search_hotel(hotel_name)
        elif menu_input == 3:
            hotel_name = input("Enter hotel name to check room availability: ").strip()
            try:
                rooms_needed = int(input("Enter number of rooms needed: "))
                room_inventory.check_room_availability(hotel_name, rooms_needed)
            except ValueError:
                print("Invalid Entry. Please enter a numerical value for rooms needed.")
        elif menu_input == 4:
            room_inventory.generate_room_inventory_report()
        elif menu_input == 5:
            print("Exiting the Room Inventory Dashboard.")
            break
        else:
            print("Invalid Entry. Please enter a number between 1 and 5.")

# Run the main menu
if __name__ == "__main__":
    main_menu()
""
```

```python
""
import pandas as pd  # Import the Pandas library for data manipulation
import matplotlib.pyplot as plt

class WeatherDashboard:
    def __init__(self):
        # Load the weather data from CSV
        self.weather_df = pd.read_csv('Weather report.csv')

    def display_weather_report(self):
        # Display the first few rows of the DataFrame to show the weather report
        print("Weather Report Data (First 5 rows):")
        print(self.weather_df.head())

    def build_weather_report(self, weather_type):
        # Generate a weather report based on the selected type
        if weather_type == "Current":
            print("Current Weather Report:")
            print(self.weather_df[['Date', 'Temperature', 'Conditions']].tail(1))
        elif weather_type == "Forecast":
            print("Weather Forecast for the Upcoming Days:")
            print(self.weather_df[['Date', 'Temperature', 'Conditions']].tail(5))
        elif weather_type == "Monthly":
            month = input("Enter the month (e.g., 'January', 'February', etc.): ").strip().capitalize()
            monthly_data = self.weather_df[self.weather_df['Month'] == month]
            if not monthly_data.empty:
                print(f"Monthly Weather Report for {month}:")
                print(monthly_data[['Date', 'High Temp', 'Low Temp', 'Conditions']])
            else:
                print(f"No data available for {month}.")
        elif weather_type == "Historical":
            print("Historical Weather Report (Daily):")
            print(self.weather_df[['Date', 'Temperature', 'Conditions']])

    def high_low_temperature_report(self, month):
        # Display the high and low temperatures for the given month
        monthly_data = self.weather_df[self.weather_df['Month'] == month]
        if not monthly_data.empty:
            high_temp = monthly_data['High Temp'].max()
            low_temp = monthly_data['Low Temp'].min()
            print(f"{month} High Temperature: {high_temp} °C")
            print(f"{month} Low Temperature: {low_temp} °C")
        else:
            print(f"No data available for {month}.")

# Main menu function to link to weather functions
def main_menu():
    # Initialize the WeatherDashboard class
    weather_dashboard = WeatherDashboard()

    while True:
        print("Weather Dashboard Main Menu")
        print("1. Display Weather Report")
        print("2. Build Weather Report")
        print("3. High and Low Temperature Report")
        print("4. Exit")

        try:
            menu_input = int(input("Please enter menu selection (1 - 4): "))
        except ValueError:
            print("Invalid Entry. Please enter a number between 1 and 4.")
            continue

        if menu_input == 1:
            weather_dashboard.display_weather_report()
        elif menu_input == 2:
            print("Select Weather Type to Build Report:")
            print("1. Current Weather")
            print("2. Forecast Weather")
            print("3. Monthly Weather")
            print("4. Historical Weather")
            try:
                weather_type_input = int(input("Please enter selection (1 - 4): "))
                if weather_type_input == 1:
                    weather_dashboard.build_weather_report("Current")
                elif weather_type_input == 2:
                    weather_dashboard.build_weather_report("Forecast")
                elif weather_type_input == 3:
                    weather_dashboard.build_weather_report("Monthly")
                elif weather_type_input == 4:
                    weather_dashboard.build_weather_report("Historical")
                else:
                    print("Invalid Entry. Please enter a number between 1 and 4.")
            except ValueError:
                print("Invalid Entry. Please enter a number between 1 and 4.")
        elif menu_input == 3:
            month = input("Enter the month for high/low temperature report (e.g., 'January', 'February', etc.): ").strip().capitalize()
            weather_dashboard.high_low_temperature_report(month)
        elif menu_input == 4:
            print("Exiting the Weather Dashboard.")
            break
        else:
            print("Invalid Entry. Please enter a number between 1 and 4.")

        # Ask if the user wants to check another weather type
        back_to_menu = input("Do you want to check another weather type? (yes/no): ").strip().lower()
        if back_to_menu != 'yes':
            print("Returning to the Main Menu.")
            break

# Run the main menu
if __name__ == "__main__":
    main_menu()
""
```
